Utiliza las secciones pertinentes para colocar la solución a cada uno de los requerimientos solicitados.

##Importa las librerías pertinentes para solucionar el reto.

```python
import pandas as pd
import numpy as np
```

##Extrae la información del archivo.

```python
# Abrir archivo
hospitalData = pd.read_excel('/content/hospitalData.xlsx')
# desplega
hospitalData
```

##Muestra la estructura y los tipos de datos de cada columna para identificar qué operaciones puedes realizar con cada una de ellas, asegúrandote de que las columnas con fechas sean del tipo datetime64.

```python
# Visualiza los tipos de datos
hospitalData.info()
# Cambia el tipo de dato de object a datetime64
hospitalData['Week Ending Date'] = pd.to_datetime(hospitalData['Week Ending Date'])
# Visualiza los tipos de datos, se observa el cambio de tipo de dato
hospitalData.info()
print("El tipo de dato es:", hospitalData['Week Ending Date'].dtype)
```

##Elimina los registros con el código geográfico MP.

```python
hospitalData.shape
# Conteo previo para verificar registros a eliminar
registros_mp_antes = (hospitalData['Geographic aggregation'] == 'MP').sum()
print(f"Registros con código 'MP' antes de eliminar: {registros_mp_antes}")

# Eliminación de los registros solicitados
hospitalData = hospitalData[hospitalData['Geographic aggregation'] != 'MP'].copy()

# conteo despues de eleminar en 0
registros_mp_despues = (hospitalData['Geographic aggregation'] == 'MP').sum()
print(f"Registros con código 'MP' después de eliminar: {registros_mp_despues}")
# Muestra el nuevo tamaño
print(f"Nuevas dimensiones del DataFrame: {hospitalData.shape}")

```

#Para las columnas que tienen que ver con la ICU, cambia los valores faltantes por 0.

```python
# Columnas con ICU
columnas_icu = ['Number of ICU Beds','Number of Adult ICU Beds','Number of Pediatric ICU Beds','Number of ICU Beds Occupied',
                'Number of Adult ICU Beds Occupied','Number of Pediatric ICU Beds Occupied']
# Cambiamos los vacios por 0
hospitalData[columnas_icu] = hospitalData[columnas_icu].fillna(0)

# Mostramos el DataFrame
#hospitalData.head(2214) ## Para vizualizar el dato 2248 y2252
hospitalData
```

#Conoce el periodo de registro que se encuentra en el archivo para saber cuántos días han transcurrido entre la fecha más reciente y la más antigua.

```python
#Vizualiza la fecha mas antigua
print(f'La fecha más antigua es: {hospitalData["Week Ending Date"].min()}')
#Vizualiza la fecha mas reciente
print(f'La fecha más reciente es: {hospitalData["Week Ending Date"].max()}')
#Resta las fechas
dias_Transcurridos = hospitalData['Week Ending Date'].max() - hospitalData['Week Ending Date'].min()
#Muestra los dias transcurridos
print(f'Los días transcurridos son: {dias_Transcurridos.days}')
```

#Obtén la cantidad total de pacientes pediátricos hospitalizados con COVID-19.

```python
#Suma la cantidad de pacientes
totalPediatricos = hospitalData['Number of Pediatric Patients Hospitalized with COVID-19'].sum()
#Visualiza el total
print(f'La cantidad total de pacientes pediátricos hospitalizados con COVID-19 es: {totalPediatricos}')
```

#Obtén el promedio de pacientes adultos hospitalizados por RSV.

```python
#Calcula el promedio
promedioAdultos = hospitalData['Number of Adult Patients Hospitalized with RSV'].mean()
#Visualiza el promedio
print(f'El promedio de pacientes adultos hospitalizados por RSV es: {promedioAdultos:.4f}')
```

#Agrupa por región geográfica y obtén la sumatoria de pacientes adultos hospitalizados con influenza.

```python
#Se seleccionan las columnas, se agrupa y se suman
AgrupacionGeograficaInflu = hospitalData[['Number of Adult Patients Hospitalized with Influenza','Geographic aggregation']].groupby('Geographic aggregation').sum().astype(int)
#Se visualiza la tabla
AgrupacionGeograficaInflu
```

#Crea una columna derivada llamada "Percent of Adult ICU Beds Occupied" con el porcentaje de ocupación para las camas de los adultos en la ICU.

```python
#Agrega la columna y se calcula el porcentaje
hospitalData['Percent of Adult ICU Beds Occupied'] = (hospitalData['Number of Adult ICU Beds Occupied']*100 / hospitalData['Number of Adult ICU Beds']).round(2)
#se visualiza la columna agregada
hospitalData
```

#Crea un *dataframe* llamado "filtro", con las filas de los hospitales que tengan más de 300 camas "disponibles" para adultos en la ICU; y las columnas de identificación geográfica y las relacionadas, con la cantidad de pacientes con alguna de las enfermedades respiratorias de los adultos.

```python
#Calculamos la cantidad de camas disponibles
disponibles = hospitalData['Number of Adult ICU Beds'] -  hospitalData['Number of Adult ICU Beds Occupied']

#Columnas que visualizaremos
columnas_EnfermedadesAdultos = [
    'Geographic aggregation',
    'Number of Adult Patients Hospitalized with COVID-19',
    'Number of Adult Patients Hospitalized with Influenza',
    'Number of Adult Patients Hospitalized with RSV'
]
#Seleccionamos las filas con mas de 300 camas disponibles y las columnas
filtro = hospitalData.loc[disponibles > 300, columnas_EnfermedadesAdultos]
filtro
```

#Crea un *dataframe* llamado "registros2024" que tenga todos los registros del año 2024.

```python
#Seleccionamos los registros del 2024 y los guardamos en registros2024
registros2024 = hospitalData[hospitalData['Week Ending Date'].dt.year == 2024]
#Mostramos
registros2024
```

#Crear una columna derivada con el conteo de la cantidad de certificaciones que tiene cada hospital; es decir, cuenta la cantidad de elementos que hay en la columna de certificaciones separándolas con /

```python
#Cuenta la cantidad de certificaciones separadas por /
cantidadCertificaciones = hospitalData['Certifications'].str.split('/').str.len()
#Agrega la columna pasando la cantidad de certificaciones
hospitalData['Cantidad de Certificaciones'] = cantidadCertificaciones
#Muestra
hospitalData
```

#Busca todos los hospitales que cumplan con la certificación CRN y déjalos en un *dataframe* llamado "CRNData".

```python
#Busca en la columna certificacion CRN y selecciona toda la fila
CRNData = hospitalData[hospitalData['Certifications'].str.contains('CRN',case=False, na=False)]
#Muestra
CRNData
```

#Almacena el *dataframe* modificado y los *dataframes* creados en los pasos anteriores en un mismo archivo de Excel llamado "datosHospitalesProcesados.xlsx". Cada *dataframe* deberá estar en una hoja diferente.

```python
#Se crea el archivo
with pd.ExcelWriter('/content/datosHospitalesProcesados.xlsx') as writer:

  #Se crean las hojas en el archivos
  hospitalData.to_excel(writer, sheet_name='General_Modificado', index=False)
  filtro.to_excel(writer, sheet_name='Filtro_Camas_ICU', index=False)
  registros2024.to_excel(writer, sheet_name='Registros_2024', index=False)
  CRNData.to_excel(writer, sheet_name='Certificaciones_CRN', index=False)

#Se imprime el mensaje que se ha creado
print("¡El archivo 'datosHospitalesProcesados.xlsx' ha sido creado y exportado exitosamente con todas sus hojas!")
```
